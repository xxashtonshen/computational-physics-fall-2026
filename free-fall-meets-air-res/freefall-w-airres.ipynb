```python
import numpy as np
import matplotlib.pyplot as plt

from typing import Final

dt = 0.033
#intial conditions
y = 1
v = 0
t = 0
g = 9.81

#groups for plotting
times = [t]
positions = [y]
velocities = [v]

while y > 0:
  v = v + g * dt
  y = y - v * dt
  t = t + dt

  times.append(t)
  velocities.append(v)
  positions.append(y)

plt.plot(times, positions)
plt.xlabel("time (s)")
plt.ylabel("positions (m)")
plt.show()
```
